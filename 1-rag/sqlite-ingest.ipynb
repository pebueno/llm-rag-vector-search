# Lesson 9 — Data Ingestion (sqlite)

Separate ingestion process: fetch data once, write to persistent `faq.db`.
The query notebook reads from it independently — no re-indexing on restart.

```python
from ingest import load_faq_data

documents = load_faq_data()
print(f"Loaded {len(documents)} documents")
```

```python
# Filter to LLM Zoomcamp only for this index
docs_llm = [doc for doc in documents if doc["course"] == "llm-zoomcamp"]
print(f"LLM Zoomcamp: {len(docs_llm)} documents")
```

```python
import time
from sqlitesearch import TextSearchIndex

index = TextSearchIndex(
    text_fields=["question", "section", "answer"],
    keyword_fields=["course"],
    db_path="faq.db"
)

# Add documents one by one — simulates real ingestion pipeline
for doc in docs_llm:
    index.add(doc)
    print(f"Added: {doc['question'][:60]}...")
    time.sleep(0.1)  # reduced from 0.5s for speed

index.close()
print("Done. Index saved to faq.db")
```
